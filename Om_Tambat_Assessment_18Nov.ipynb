```python
#Section A : Supervised Learning

#Q1. Logistic Regression on the wine dataset

#(a)
from sklearn.datasets import load_wine
import pandas as pd
import numpy as np

X, y = load_wine(return_X_y=True, as_frame=True)

print("Shape of X:", X.shape)
print("\nClass counts:")
print(y.value_counts().sort_index())

#Explanation:
# X contains input features.
# y contains target classes.
# X.shape displays rows and columns.
# value_counts() displays the number of samples in each class.

```

```python
#(b) Split the dataset


from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.3,
    random_state=42
)

print("Training shape:", X_train.shape)
print("Testing shape:", X_test.shape)


#Explanation:
# 70% data for training.
# 30% data for testing.
# random_state=42 makes the split reproducible.
```

```python
#(c) Scale the data


from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Training data scaled successfully")
print("Testing data scaled successfully")

#Explanation:
# fit_transform() learns the scaling values from training data and scales it.
# transform() applies the same scaling to test data.
# This prevents data leakage.
```

```python
#(d) Train Logistic Regression


from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

model = LogisticRegression(max_iter=1000)

model.fit(X_train_scaled, y_train)

y_pred = model.predict(X_test_scaled)

print("Test Accuracy:", accuracy_score(y_test, y_pred))

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
Precision explanation (Text cell)

Precision for class 0 means out of all the samples predicted as class 0, how many actually belong to class 0.
```

```python
#Q2. One Tree Against a Forest

#(a) Decision Tree

from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

dt = DecisionTreeClassifier(random_state=42)

dt.fit(X_train, y_train)

dt_pred = dt.predict(X_test)

dt_accuracy = accuracy_score(y_test, dt_pred)

print("Decision Tree Accuracy:", dt_accuracy)
```

```python
#(b) Random Forest


from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

rf.fit(X_train, y_train)

rf_pred = rf.predict(X_test)

rf_accuracy = accuracy_score(y_test, rf_pred)

print("Random Forest Accuracy:", rf_accuracy)
```

```python
#(c) Top 3 important features


importance = pd.Series(
    rf.feature_importances_,
    index=X.columns
)

top_features = importance.sort_values(ascending=False).head(3)

print("Top 3 Important Features:")
print(top_features)
```

```python
#(d) Explanation

Random Forest usually performs better than a single Decision Tree because it combines multiple trees and reduces overfitting by averaging their predictions.
```

```python
#Q3. KNN and the Scaling Trap

#(a)KNN on the unscaled training data

from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

knn1 = KNeighborsClassifier(n_neighbors=5)

knn1.fit(X_train, y_train)

pred1 = knn1.predict(X_test)

accuracy1 = accuracy_score(y_test, pred1)

print("Accuracy on the unscaled training data:", accuracy1)
```

```python
#(b) KNN on the scaled training data


knn2 = KNeighborsClassifier(n_neighbors=5)

knn2.fit(X_train_scaled, y_train)

pred2 = knn2.predict(X_test_scaled)

accuracy2 = accuracy_score(y_test, pred2)

print("Accuracy on the scaled training data:", accuracy2)
```

```python
#(c) Print two numbers together on one line so the difference is visible


print("Accuracy on the unscaled training data:", accuracy1)
print("Accuracy on the scaled training data", accuracy2)
```

```python
#(d) Explanation

Scaling changes KNN results because KNN calculates distances between data points, and large-scale features can dominate those distances. Decision Trees use feature-based splitting rather than distance calculations, so scaling usually has little effect on them.
```

```python
#Section B : Unsupervised Learning

#Q4. K-Means Clustering

#(a) Scale all X data

from sklearn.preprocessing import StandardScaler

scaler_kmeans = StandardScaler()

X_scaled = scaler_kmeans.fit_transform(X)

print("Original Data Shape:", X.shape)
print("Scaled Data Shape:", X_scaled.shape)
```

```python
#((b) Apply K-Means


from sklearn.cluster import KMeans

kmeans = KMeans(
    n_clusters=3,
    n_init=10,
    random_state=42
)

labels = kmeans.fit_predict(X_scaled)

print("Cluster Labels:")
print(labels)
```

```python
#(c) Count rows in each cluster


print("Number of rows in each cluster:")

print(pd.Series(labels).value_counts().sort_index())
```

```python
#(d) Attach cluster labels to DataFrame


df = X.copy()

df["cluster"] = labels

print("DataFrame with Cluster:")
print(df.head())

# Calculate average values


cluster_means = df.groupby("cluster")[
    ["alcohol", "proline", "color_intensity"]
].mean()

print("Cluster Average Values:")
print(cluster_means)


```

```python
#(b) Reads the rows with csv.DictReader

def read_marks(path):

    try:
        # Open the CSV file
        with open(path, "r") as file:

            # Read CSV rows
            reader = csv.DictReader(file)

            total_rows = 0
            marks_list = []

            for row in reader:

                total_rows += 1

                marks = row["marks"].strip()

                # Skip empty marks
                if marks == "":
                    continue

                marks_list.append(float(marks))

            # Calculate count and average
            count = len(marks_list)

            average = sum(marks_list) / count if count > 0 else 0

            # Create dictionary
            result = {
                "total_rows": total_rows,
                "rows_with_marks": count,
                "average_marks": average
            }

            # Write dictionary into JSON
            with open("summary.json", "w") as json_file:
                json.dump(result, json_file, indent=4)

            print("Summary Dictionary:")
            print(result)

            # Read JSON file
            print("\nReading summary.json:")

            with open("summary.json", "r") as json_file:
                print(json.load(json_file))

            return result

    except FileNotFoundError:
        print(f"File not found: {path}")
        return None
```

```python
#(c) Skips any row whose marks value is empty, then returns a dictionary with the total number of rows, how many had marks, and the average marks.

# First call
print("First Function Call:")
read_marks("students.csv")

# Second call
print("\nSecond Function Call:")
read_marks("missing.csv")
```

```python
#(d)  Write dictionary to summary.json using JSON module

import csv
import json

def read_marks(path):
    try:
        with open(path, "r") as file:
            reader = csv.DictReader(file)

            summary = {}

            for row in reader:
                name = row["name"]
                marks = row["marks"]

                if marks:
                    summary[name] = int(marks)
                else:
                    summary[name] = None

        # Write dictionary to JSON file
        with open("summary.json", "w") as file:
            json.dump(summary, file, indent=4)

        # Read JSON file and print
        with open("summary.json", "r") as file:
            print(json.load(file))

    except FileNotFoundError:
        print(f"File not found: {path}")
        return None


# Call function twice
read_marks("students.csv")
read_marks("missing.csv")
```
